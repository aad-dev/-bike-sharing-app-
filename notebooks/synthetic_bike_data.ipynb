```python
import pandas as pd
import numpy as np

# instellen seed voor reproduceerbaarheid
np.random.seed(42)

# genereren van 150 fictieve rijen 
n = 150

data = {
    'season': np.random.choice([1, 2, 3, 4], n),
    'yr': np.random.choice([0, 1], n),
    'mnth': np.random.randint(1, 13, n),
    'hr': np.random.randint(0, 24, n),
    # feestdagen zijn zeldzaam, dus geven we '0' een kans van 95%
    'holiday': np.random.choice([0, 1], n, p=[0.95, 0.05]),
    'weekday': np.random.randint(0, 7, n),
    # werk dagen komen vaker voor (na een berekening: ongeveer 71% van de week)
    'workingday': np.random.choice([0, 1], n, p=[0.29, 0.71]),
    'weathersit': np.random.choice([1, 2, 3, 4], n, p=[0.6, 0.3, 0.08, 0.02]),
    # numerieke waarden waren in de originele dataset genormaliseerd tussen 0 en 1
    'temp': np.round(np.random.uniform(0.1, 0.9, n), 2),
    'atemp': np.round(np.random.uniform(0.1, 0.9, n), 2),
    'hum': np.round(np.random.uniform(0.2, 0.9, n), 2),
    'windspeed': np.round(np.random.uniform(0.0, 0.6, n), 2)
}

# aanmaken dataframe en opslaan als csv
df_synthetic = pd.DataFrame(data)
df_synthetic.to_csv('synthetic_bike_data.csv', index=False)

print("Synthetische dataset succesvol gegenereerd en opgeslagen als 'synthetic_bike_data.csv'")
display(df_synthetic.head())
```
